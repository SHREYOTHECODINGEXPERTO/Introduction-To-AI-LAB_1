```python
#problem3
# Road network using adjacency list
# A = Kolkata, B = Durgapur, C = Siliguri
# D = Asansol, E = Burdwan, F = Delhi, G = Jaipur

graph = {
    'Kolkata': ['Durgapur', 'Siliguri'],
    'Durgapur': ['Kolkata', 'Asansol'],
    'Siliguri': ['Kolkata'],
    'Asansol': ['Durgapur', 'Burdwan'],
    'Burdwan': ['Asansol'],
    'Delhi': ['Jaipur'],
    'Jaipur': ['Delhi']
}


# DFS to check route between two cities
def has_route(graph, start, destination, visited=None):
    if visited is None:
        visited = set()

    if start == destination:
        return True

    visited.add(start)

    for city in graph[start]:
        if city not in visited:
            if has_route(graph, city, destination, visited):
                return True

    return False


# DFS to find one connected component
def find_component(city, visited):
    component = []
    visited.add(city)
    component.append(city)

    for neighbour in graph[city]:
        if neighbour not in visited:
            component += find_component(neighbour, visited)

    return component


# Check route
start = 'Kolkata'
destination = 'Burdwan'

if has_route(graph, start, destination):
    print(f"Route exists from {start} to {destination}")
else:
    print(f"No route exists from {start} to {destination}")


# Find connected components
visited = set()
components = []

for city in graph:
    if city not in visited:
        components.append(find_component(city, visited))

print("\nConnected Components:")
for i, component in enumerate(components, 1):
    print(f"Component {i}: {component}")
```
